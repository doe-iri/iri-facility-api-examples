# Globus (ALCF) → IRI Token Bootstrap Notebook

This notebook performs an interactive OAuth2 flow using **Globus Auth** to authenticate with ALCF.

It will:
* Start an OAuth authorization flow
* Exchange the authorization code for an access token
* Save the "iri_api"-scoped token to `~/.iri_token.json` as `IRI_API_TOKEN`


## 1) Setup and Imports

```python
import json
import sys
from pathlib import Path

# Install `alcf-tokens` 
!{sys.executable} -m pip install alcf-tokens
```

## 2) Create OAuth Flow

```python
from alcf_tokens.auth import login

# Login to the ALCF IRI service
# Click on the URL, authenticate with your ALCF credentials, 
# and copy-paste the authorization code in the text box below
login("iri")
```

```python
from alcf_tokens.cli import test_token

# Test your token to see if it is ready to use
# This should give your {"ready": true, "error": null}
test_token("iri")
```

## 3) Save Token for Other Notebooks

```python
from alcf_tokens.auth import get_access_token

iri_token = get_access_token("iri")

home_path = Path("~/.iri_token.json").expanduser()
home_path.parent.mkdir(parents=True, exist_ok=True)

with open(home_path, "w") as f:
    json.dump({"IRI_API_TOKEN": get_access_token("iri")}, f)
print(f"Token saved: {home_path}")
```
