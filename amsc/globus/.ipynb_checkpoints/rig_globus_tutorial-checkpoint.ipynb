# RIG + Globus Transfer — end-to-end tutorial (via KongPing -> RIG)

<div style="border:2px solid #c00;border-radius:8px;padding:10px;color:#c00;font-weight:bold">
Sandbox / demo notebook. It drives the huck local sandbox (Kong -> RIG -> Globus Transfer).
It performs REAL Globus operations against endpoint <code>44b05bb7-576b-4bf1-a3f6-df0a7a13f9b7</code>.
Do not point it at a production RIG.
</div>

This notebook shows the full data-movement path **through the AmSC platform** — your code never
touches a Globus token. You send an **AmSC (PingAM) bearer**; Kong validates it and injects
`X-Userinfo`; RIG looks up your linked Globus credential in its Tier-2 vault and substitutes it
before forwarding to `transfer.api.globus.org`.

**Steps**

1. Get an AmSC token (from `huck/scripts/get-project-token.sh`)
2. Link your Globus account + consent (RIG Tier-2 OOB flow)
3. List files
4. Copy a file (dirA -> dirB on the same endpoint)
5. Delete the copied file

Every call in steps 2-5 goes to RIG at `https://localhost.rig.american-science-cloud.org`.

---

### Prerequisites

- The **huck sandbox is up**: `cd amsc-gitlab/huck && make start` (Kong, RIG, Vault all running).
- `/etc/hosts` has `127.0.0.1 localhost.rig.american-science-cloud.org` (done by `make setup`).
- `mkcert` installed (the notebook auto-detects its root CA for local TLS).
- An AmSC token file from `huck/scripts/get-project-token.sh` (see Step 1) — no client secret needed here.
- A Globus account that can access endpoint `44b05bb7-...`, plus a test file on it for the copy step.
- `pip install requests` (and `mkcert` on PATH).

## 0) Configuration

All knobs are env-overridable. Defaults match the huck sandbox and
`amsc-gitlab/huck/scripts/get-project-token.sh`.

Note on TLS: RIG runs behind a local mkcert cert, so RIG calls verify against the mkcert root CA
(auto-detected via `mkcert -CAROOT`). For dev only you can set `RIG_SKIP_TLS_VERIFY=1`.

**Choose which endpoint to test** with `GLOBUS_SITE` (or edit `SELECTED` in the config cell): one of
`parent`, `esnet`, `nersc`, `alcf`, `personal` — the same set as the smoke test. Default `parent`
(`44b05bb7-...`). Point at any other endpoint with `GLOBUS_ENDPOINT=<uuid>`. The `ENDPOINTS` table records which
collections each endpoint needs pre-consented, so GCS v5 endpoints (`parent`, `alcf`) auto-request
`data_access` at link time — no ConsentRequired round-trip. Each entry also carries a default
base `path` (`/juztas` for `alcf`, `/~/` for the rest) used by `ls` and the copy step; override
with `*_BASE_PATH` (e.g. `ALCF_BASE_PATH`) or `LS_PATH`/`SRC_DIR`/`DST_DIR`.

````python
import os
import json
import time
import base64
import subprocess

import requests
import urllib3

# ---- RIG / sandbox ----
RIG_BASE_URL = os.environ.get(
    "RIG_BASE_URL", "https://localhost.rig.american-science-cloud.org"
).rstrip("/")
FACILITY = "globus-transfer"

# ---- Globus endpoints you can test (same set as smoketest/globus_transfer_smoke.py) ----
# "data_access": collection UUIDs that MUST be pre-consented at link time (GCS v5 mapped/
# guest collections). Pre-consenting an endpoint that does NOT expose a data_access scope
# (GCP personal, GCS non-v5) breaks the Globus auth request, so only list UUIDs known to
# need it. Empty list -> link with transfer:all only. Requirements per the smoke test +
# observed ConsentRequired responses.
ENDPOINTS = {
    "parent":   {"uuid": os.environ.get("PARENT_ENDPOINT", "44b05bb7-576b-4bf1-a3f6-df0a7a13f9b7"),
                 "label": "GCS parent endpoint (GCS v5; hosts the esnet guest)",
                 "data_access": ["44b05bb7-576b-4bf1-a3f6-df0a7a13f9b7"],  # confirmed: ConsentRequired on own UUID
                 "path": os.environ.get("PARENT_BASE_PATH", "/~/home/jbalcas/")},  # demo user home; override per user
    "esnet":    {"uuid": os.environ.get("ESNET_EP", "48b5b912-4cfc-40b7-b6ee-a8f50f0e6c8a"),
                 "label": "ESnet guest collection (on the parent endpoint)",
                 "data_access": [],   # guest on non-v5 parent: transfer:all only
                 "path": os.environ.get("ESNET_BASE_PATH", "/~/")},
    "nersc":    {"uuid": os.environ.get("NERSC_EP", "9d6d994a-6d04-11e5-ba46-22000b92c6ec"),
                 "label": "NERSC DTN (official GCS)",
                 "data_access": [],   # v4/official: transfer:all (relink via consent if v5)
                 "path": os.environ.get("NERSC_BASE_PATH", "/~/")},
    "alcf":     {"uuid": os.environ.get("ALCF_EP", "9032dd3a-e841-4687-a163-2720da731b5b"),
                 "label": "ALCF Eagle (GCS v5)",
                 "data_access": ["9032dd3a-e841-4687-a163-2720da731b5b"],  # GCS v5: needs data_access
                 "path": os.environ.get("ALCF_BASE_PATH", "/juztas")},
    "personal": {"uuid": os.environ.get("PERSONAL_EP", "7ff63733-2dbf-11f1-a24c-02535127e3d7"),
                 "label": "GCP personal (Globus Connect Personal)",
                 "data_access": [],   # GCP: transfer:all only, never data_access
                 "path": os.environ.get("PERSONAL_BASE_PATH", "/~/")},
}

# Pick which endpoint to drive (env GLOBUS_SITE, default "parent" = 44b05bb7-...).
# Or override with GLOBUS_ENDPOINT=<uuid> to point at something not listed above.
SELECTED = os.environ.get("GLOBUS_SITE", "parent")
if SELECTED not in ENDPOINTS:
    raise ValueError(f"GLOBUS_SITE must be one of {list(ENDPOINTS)} (got {SELECTED!r})")
ENDPOINT = os.environ.get("GLOBUS_ENDPOINT", ENDPOINTS[SELECTED]["uuid"])
# Collections this endpoint needs pre-consented (override via PRECONSENT_COLLECTIONS).
DATA_ACCESS_COLLECTIONS = list(ENDPOINTS[SELECTED]["data_access"])
ENDPOINT_NEEDS_DATA_ACCESS = bool(DATA_ACCESS_COLLECTIONS)
ENDPOINT_PATH = ENDPOINTS[SELECTED]["path"]   # default ls / copy base dir for this endpoint

# ---- AmSC token ----
# This notebook does NOT log you in. Generate a project-scoped token first with
# huck/scripts/get-project-token.sh (see Step 1), then point AMSC_TOKEN_FILE at the file
# it writes (default /tmp/pingam_project_token.txt), or export AMSC_PROJECT_TOKEN.
AMSC_TOKEN_FILE = os.environ.get("AMSC_TOKEN_FILE", "/tmp/pingam_project_token.txt")


def _resolve_rig_verify():
    # requests verify= for RIG calls: False, a CA bundle path, or True. Auto-detects mkcert CAROOT.
    if os.environ.get("RIG_SKIP_TLS_VERIFY", "").lower() in ("1", "true", "yes"):
        urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
        print("  ! RIG TLS verification disabled (RIG_SKIP_TLS_VERIFY)")
        return False
    ca = os.environ.get("RIG_CA_BUNDLE", "")
    if ca:
        return os.path.expanduser(os.path.expandvars(ca))
    try:
        result = subprocess.run(["mkcert", "-CAROOT"], capture_output=True, text=True, timeout=2)
        if result.returncode == 0:
            ca_path = os.path.join(result.stdout.strip(), "rootCA.pem")
            if os.path.isfile(ca_path):
                return ca_path
    except Exception:
        pass
    return True


RIG_VERIFY = _resolve_rig_verify()   # local mkcert cert

# Mutable token holder, filled by Step 1.
TOKEN = {"access_token": "", "refresh_token": ""}


def rig_request(method, path, **kwargs):
    # Call RIG through Kong with the AmSC bearer. Kong validates the token and injects X-Userinfo,
    # from which RIG derives the user sub for the Tier-2 vault lookup. Tier-2 needs no X-AmSC-Project.
    headers = kwargs.pop("headers", {})
    if TOKEN["access_token"]:
        headers["Authorization"] = f"Bearer {TOKEN['access_token']}"
    return requests.request(method, f"{RIG_BASE_URL}{path}", headers=headers,
                            verify=RIG_VERIFY, **kwargs)


print(f"RIG:      {RIG_BASE_URL}")
print(f"Facility: {FACILITY}")
print(f"Site:     {SELECTED}  ({ENDPOINTS[SELECTED]['label']})")
print(f"Endpoint: {ENDPOINT}")
print(f"data_access pre-consent: {DATA_ACCESS_COLLECTIONS or 'none'}")
print(f"Base path: {ENDPOINT_PATH}")
print(f"Token:    {AMSC_TOKEN_FILE}")
print(f"RIG verify: {RIG_VERIFY}")
````

## 1) Get an AmSC token

This notebook does **not** perform the login. Generate a project-scoped AmSC token first with the
huck helper, then point this notebook at the file it writes.

```bash
cd amsc-gitlab/huck
PROJECT=PRJ-virtual-organization ./scripts/get-project-token.sh
```

It opens a device-code approval URL in your browser and, once approved, writes
`{"access_token": ..., "refresh_token": ...}` to `/tmp/pingam_project_token.txt` (override with
`OUT_TOKEN_FILE`). The token's `aud` already covers the facility APIs and its project context is
embedded by the script.

Then either set `AMSC_TOKEN_FILE` to that path (default `/tmp/pingam_project_token.txt`) or export
`AMSC_PROJECT_TOKEN`, and run the next cell. Kong validates this bearer on every RIG call.

````python
# Load the AmSC token generated by huck/scripts/get-project-token.sh (see the step above).
# Priority: AMSC_PROJECT_TOKEN env var, else the JSON file at AMSC_TOKEN_FILE.
env_token = os.environ.get("AMSC_PROJECT_TOKEN", "")
if env_token:
    TOKEN["access_token"] = env_token
    TOKEN["refresh_token"] = os.environ.get("AMSC_PROJECT_REFRESH_TOKEN", "")
    print("Loaded AmSC token from AMSC_PROJECT_TOKEN.")
else:
    if not os.path.isfile(AMSC_TOKEN_FILE):
        raise FileNotFoundError(
            f"No token file at {AMSC_TOKEN_FILE}. Generate one first:\n"
            f"  cd amsc-gitlab/huck && PROJECT=<your-project> ./scripts/get-project-token.sh\n"
            f"then set AMSC_TOKEN_FILE to its path (default /tmp/pingam_project_token.txt)."
        )
    with open(AMSC_TOKEN_FILE) as fh:
        data = json.load(fh)
    TOKEN["access_token"] = data.get("access_token", "")
    TOKEN["refresh_token"] = data.get("refresh_token", "")
    print(f"Loaded AmSC token from {AMSC_TOKEN_FILE}.")

if not TOKEN["access_token"]:
    raise RuntimeError("No access_token found — re-run get-project-token.sh and check the file.")


def _claims(jwt):
    seg = jwt.split(".")[1]
    seg += "=" * (-len(seg) % 4)
    return json.loads(base64.urlsafe_b64decode(seg))


claims = _claims(TOKEN["access_token"])
print("sub:                 ", claims.get("sub"))
print("amsc_project_context:", claims.get("amsc_project_context", "(not in token; see script output)"))
print("expires:             ", time.strftime("%Y-%m-%d %H:%M:%S", time.localtime(claims.get("exp", 0))))
````

## 2) Link your Globus account + consent (RIG Tier-2 OOB flow)

RIG stores a per-user Globus token so the data-movement layer never handles one directly. Linking
is a one-time **out-of-band PKCE** flow:

- `GET /tier2/globus-transfer/auth-url` -> RIG returns a Globus `auth_url` + a `code_verifier`.
- You open `auth_url`, authenticate at Globus, and copy the authorization code shown on the page.
- `POST /tier2/globus-transfer/exchange {code, code_verifier}` -> RIG exchanges + vaults the token.

**Scope / consent note.** The base scope is `transfer:all` (+ `offline_access` so RIG can refresh
silently). That is enough for GCP personal endpoints and GCS non-v5 collections. For a **GCS v5
mapped/guest collection**, the endpoint also needs a `data_access` scope — request it at link time
by setting `PRECONSENT` to the collection UUID. Passing a non-v5 / GCP UUID there makes Globus
reject the request with "unknown scopes", so when in doubt **leave it empty** and rely on the
`ConsentRequired -> re-link` flow demonstrated in Step 3. This notebook auto-sets `PRECONSENT` to the
selected endpoint to the `data_access` collections from the `ENDPOINTS` table (e.g. `parent`, `alcf`),
and leaves it empty otherwise.

````python
# Pre-consent the data_access collections this endpoint needs (from the ENDPOINTS table).
# Override with PRECONSENT_COLLECTIONS=<uuid,...>, or "" to force transfer:all only.
default_preconsent = ",".join(DATA_ACCESS_COLLECTIONS)
PRECONSENT = os.environ.get("PRECONSENT_COLLECTIONS", default_preconsent)

params = {"collections": PRECONSENT} if PRECONSENT else None
r = rig_request("GET", f"/tier2/{FACILITY}/auth-url", params=params)
r.raise_for_status()
link = r.json()
CODE_VERIFIER = link["code_verifier"]

print("Open this Globus URL, authenticate, and copy the authorization code:\n")
print("    " + link["auth_url"])
print("\ndata_access pre-consent requested for:", link.get("data_access_collections") or "(none)")
````

````python
code = input("Paste the Globus authorization code: ").strip()
r = rig_request("POST", f"/tier2/{FACILITY}/exchange",
                json={"code": code, "code_verifier": CODE_VERIFIER})
r.raise_for_status()
print("Linked:", json.dumps(r.json(), indent=2))

s = rig_request("GET", f"/tier2/{FACILITY}/status")
print("\nStatus:", json.dumps(s.json(), indent=2))
````

## 3) List files

A normal Globus Transfer `ls`, but routed through RIG. RIG swaps in your vaulted Globus token.

If the endpoint is a GCS v5 collection you did **not** pre-consent, Globus returns `ConsentRequired`
and RIG translates it into an RFC 9470 `globus_consent_required` 401 carrying the exact
`required_scopes`. The helper below surfaces that and tells you how to re-link (re-run Step 2 with
`PRECONSENT` set to the UUID it prints).

````python
LS_PATH = os.environ.get("LS_PATH", ENDPOINT_PATH)


def globus_ls(endpoint, path):
    r = rig_request("GET", f"/{FACILITY}/v0.10/endpoint/{endpoint}/ls", params={"path": path})
    if r.status_code == 401:
        body = r.json()
        if body.get("error") == "globus_consent_required":
            scopes = body.get("required_scopes", [])
            uuids = ",".join(
                s.split("/scopes/")[1].split("/")[0]
                for s in scopes if "/scopes/" in s
            )
            print("ConsentRequired (RFC 9470). Missing scopes:", scopes)
            print(f"-> Re-run Step 2 with PRECONSENT = '{uuids}', then retry this cell.")
            return []
        if body.get("error") == "globus_relink_required":
            print("No stored Globus credential (or it expired). Re-run Step 2 to link.")
            return []
    r.raise_for_status()
    return r.json().get("DATA", [])


items = globus_ls(ENDPOINT, LS_PATH)
print(f"\n{len(items)} item(s) under {LS_PATH}:")
for it in items[:25]:
    print(f"  {it.get('type','?'):9s} {it.get('name','?')}")
if len(items) > 25:
    print(f"  ... and {len(items) - 25} more")
````

## 4) Copy a file (dirA -> dirB on the same endpoint)

Globus transfers are asynchronous: you fetch a one-time `submission_id`, POST a `transfer` task,
then poll until it reaches `SUCCEEDED`. Here source and destination are the **same** endpoint
`44b05bb7-...`, so only the one link/consent from Step 2 is needed.

By default the copy source is the **first file listed in Step 3** under this endpoint's base path,
copied into a `rig_tutorial_copy/` subdir (Globus creates it). Override with `FILENAME`/`SRC_DIR`/`DST_DIR`.

````python
SRC_DIR = os.environ.get("SRC_DIR", ENDPOINT_PATH)
DST_DIR = os.environ.get("DST_DIR", ENDPOINT_PATH.rstrip("/") + "/rig_tutorial_copy/")
# Source file: env FILENAME, else the first regular file from the Step 3 listing.
_listed = [it.get("name") for it in globals().get("items", []) if it.get("type") == "file"]
FILENAME = os.environ.get("FILENAME", _listed[0] if _listed else "rig_tutorial_demo.txt")

src_path = SRC_DIR.rstrip("/") + "/" + FILENAME
dst_path = DST_DIR.rstrip("/") + "/" + FILENAME


def submission_id():
    r = rig_request("GET", f"/{FACILITY}/v0.10/submission_id")
    r.raise_for_status()
    return r.json()["value"]


def poll(task_id, timeout=300, interval=5):
    deadline = time.time() + timeout
    while time.time() < deadline:
        r = rig_request("GET", f"/{FACILITY}/v0.10/task/{task_id}")
        r.raise_for_status()
        b = r.json()
        st = b.get("status")
        print(f"  status={st}  bytes_transferred={b.get('bytes_transferred', 0)}")
        if st == "SUCCEEDED":
            return True
        if st in ("FAILED", "CANCELED", "EXPIRED"):
            print("  details:", b.get("nice_status_details"))
            return False
        time.sleep(interval)
    print("  timed out")
    return False


sid = submission_id()
print(f"submission_id: {sid}")
print(f"copy: {src_path}  ->  {dst_path}  (endpoint {ENDPOINT})")

r = rig_request("POST", f"/{FACILITY}/v0.10/transfer", json={
    "DATA_TYPE": "transfer",
    "submission_id": sid,
    "source_endpoint": ENDPOINT,
    "destination_endpoint": ENDPOINT,
    "DATA": [{
        "DATA_TYPE": "transfer_item",
        "source_path": src_path,
        "destination_path": dst_path,
        "recursive": False,
    }],
    "label": "RIG tutorial copy",
    "notify_on_succeeded": False,
    "notify_on_failed": True,
})
r.raise_for_status()
copy_task = r.json()["task_id"]
print(f"transfer task: {copy_task}\n")
poll(copy_task)
````

## 5) Delete the copied file

Globus delete is also an async task: fetch a `submission_id`, POST a `delete` task, poll. This
removes the destination file we just created.

````python
sid = submission_id()
print(f"submission_id: {sid}")
print(f"delete: {dst_path}  (endpoint {ENDPOINT})")

r = rig_request("POST", f"/{FACILITY}/v0.10/delete", json={
    "DATA_TYPE": "delete",
    "submission_id": sid,
    "endpoint": ENDPOINT,
    "DATA": [{"DATA_TYPE": "delete_item", "path": dst_path}],
    "label": "RIG tutorial cleanup",
})
r.raise_for_status()
del_task = r.json()["task_id"]
print(f"delete task: {del_task}\n")
poll(del_task)
````

## Recap & troubleshooting

You moved data **without ever holding a Globus token**: PingAM keycard -> Kong (validate + inject
`X-Userinfo`) -> RIG (Tier-2 vault swap + silent refresh) -> Globus Transfer.

| Symptom | Meaning | Fix |
|---|---|---|
| `globus_relink_required` (401) | No stored Globus credential, or refresh token expired | Re-run Step 2 to link |
| `globus_consent_required` (401) | GCS v5 collection needs `data_access` | Re-run Step 2 with `PRECONSENT` = the printed UUID(s) |
| 401 with `WWW-Authenticate: ... invalid_token` | AmSC/Kong rejected the bearer (expired) | Re-run Step 1 |
| `SSLError` on RIG calls | mkcert root CA not found | Install `mkcert`, or set `RIG_CA_BUNDLE`, or `RIG_SKIP_TLS_VERIFY=1` (dev only) |
| Transfer `FAILED` | Globus task-level error (e.g. source file missing) | Check `nice_status_details`; confirm `FILENAME` exists under `SRC_DIR` |

To unlink your Globus account from RIG (removes the stored Tier-2 token), uncomment and run:

````python
print(rig_request("DELETE", f"/tier2/{FACILITY}/disconnect").json())
````
